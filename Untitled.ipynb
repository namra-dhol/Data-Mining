```python
import numpy as np

# Step 1: Define vertices and edges
nodes = ["a","b","c","d","e","f","g"]
node_index = {n:i for i,n in enumerate(nodes)}
edges = [("a","b"),("b","c"),("d","a"),("a","c"),
         ("e","f"),("f","g"),("e","g"),("c","e"),("b","e")]

n = len(nodes)
A = np.zeros((n,n))

# Step 2: Build adjacency matrix
for u,v in edges:
    i, j = node_index[u], node_index[v]
    A[i,j] = 1
    A[j,i] = 1   # undirected

# Step 3: Degree matrix & Laplacian
D = np.diag(A.sum(axis=1))
L = D - A

# Step 4: Eigen decomposition
eigvals, eigvecs = np.linalg.eigh(L)

# Sort eigenvalues and corresponding vectors
idx = np.argsort(eigvals)
eigvals = eigvals[idx]
eigvecs = eigvecs[:, idx]

# Step 5: Fiedler vector (2nd smallest eigenvalue)
fiedler_vector = eigvecs[:,1]

# Cluster assignment: sign of Fiedler vector
cluster1 = [nodes[i] for i,v in enumerate(fiedler_vector) if v >= 0]
cluster2 = [nodes[i] for i,v in enumerate(fiedler_vector) if v < 0]

print("Adjacency Matrix:\n", A)
print("Laplacian Matrix:\n", L)
print("Eigenvalues:", eigvals)
print("Fiedler Vector:", fiedler_vector)
print("Cluster 1:", cluster1)
print("Cluster 2:", cluster2)
```

```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Step 1: Define vertices and edges
nodes = ["a","b","c","d","e","f","g"]
node_index = {n:i for i,n in enumerate(nodes)}
edges = [("a","b"),("b","c"),("d","a"),("a","c"),
         ("e","f"),("f","g"),("e","g"),("c","e"),("b","e")]

n = len(nodes)
A = np.zeros((n,n))

# Step 2: Build adjacency matrix
for u,v in edges:
    i, j = node_index[u], node_index[v]
    A[i,j] = 1
    A[j,i] = 1   # undirected

# Step 3: Degree matrix & Laplacian
D = np.diag(A.sum(axis=1))
L = D - A

# Step 4: Eigen decomposition
eigvals, eigvecs = np.linalg.eigh(L)

# Sort eigenvalues and vectors
idx = np.argsort(eigvals)
eigvals = eigvals[idx]
eigvecs = eigvecs[:, idx]

# Step 5: Use 2nd & 3rd eigenvectors for 4-way cut
X = eigvecs[:, 1:3]   # take eigenvector 2 and 3

# KMeans clustering
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)

# Assign nodes to clusters
clusters = {i: [] for i in range(4)}
for node, label in zip(nodes, labels):
    clusters[label].append(node)

print("Eigenvalues:", eigvals)
print("Clusters (4-way cut):")
for cid, members in clusters.items():
    print(f"Cluster {cid+1}: {members}")

# --- Visualization 1: Graph view ---
G = nx.Graph()
G.add_nodes_from(nodes)
G.add_edges_from(edges)

pos = nx.spring_layout(G, seed=42)

# Colors for up to 4 clusters
color_map = ["skyblue", "lightgreen", "salmon", "khaki"]
colors = [color_map[label] for label in labels]

plt.figure(figsize=(12,6))

# Left: Graph structure
plt.subplot(1,2,1)
nx.draw(G, pos, with_labels=True, node_color=colors, node_size=1000,
        font_size=12, font_weight="bold", edge_color="gray")
plt.title("Graph with Spectral Clusters")

# Right: 2D spectral embedding
plt.subplot(1,2,2)
for i, node in enumerate(nodes):
    plt.scatter(X[i,0], X[i,1], color=color_map[labels[i]], s=200, edgecolor="k")
    plt.text(X[i,0]+0.02, X[i,1]+0.02, node, fontsize=12, weight="bold")

plt.title("2D Spectral Embedding (2nd vs 3rd eigenvectors)")
plt.xlabel("Eigenvector 2")
plt.ylabel("Eigenvector 3")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Step 1: Define vertices and edges
nodes = ["a","b","c","d","e","f","g"]
node_index = {n:i for i,n in enumerate(nodes)}
# edges = [("a","b"),("b","c"),("d","a"),("a","c"),
#          ("e","f"),("f","g"),("e","g"),("c","e"),("b","e")]
edges = [("a", "b"), ("b", "c"), ("d", "a"), ("a", "c"), ("c","d"),
         ("e", "f"), ("f", "g"), ("e", "g"), ("c", "e"), ("b", "e")]

n = len(nodes)
A = np.zeros((n,n))

# Step 2: Build adjacency matrix
for u,v in edges:
    i, j = node_index[u], node_index[v]
    A[i,j] = 1
    A[j,i] = 1   # undirected

# Step 3: Degree matrix & Laplacian
D = np.diag(A.sum(axis=1))
L = D - A

# Step 4: Eigen decomposition
eigvals, eigvecs = np.linalg.eigh(L)

# Sort eigenvalues and vectors
idx = np.argsort(eigvals)
eigvals = eigvals[idx]
eigvecs = eigvecs[:, idx]

# Step 5: Scale eigenvectors by 1/sqrt(lambda)
# Take eigenvector 2 and 3 with their eigenvalues
v2 = eigvecs[:, 1] / np.sqrt(eigvals[1])
v3 = eigvecs[:, 2] / np.sqrt(eigvals[2])

X = np.column_stack((v2, v3))  # shape (n,2)

# KMeans clustering (4-way cut)
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)

# Assign nodes to clusters
clusters = {i: [] for i in range(4)}
for node, label in zip(nodes, labels):
    clusters[label].append(node)

print("Eigenvalues:", eigvals)
print("Clusters (4-way cut with scaling):")
for cid, members in clusters.items():
    print(f"Cluster {cid+1}: {members}")

# --- Visualization 1: Graph view ---
G = nx.Graph()
G.add_nodes_from(nodes)
G.add_edges_from(edges)

pos = nx.spring_layout(G, seed=42)

# Colors for up to 4 clusters
color_map = ["skyblue", "lightgreen", "salmon", "khaki"]
colors = [color_map[label] for label in labels]

plt.figure(figsize=(12,6))

# Left: Graph structure
plt.subplot(1,2,1)
nx.draw(G, pos, with_labels=True, node_color=colors, node_size=1000,
        font_size=12, font_weight="bold", edge_color="gray")
plt.title("Graph with Spectral Clusters")

# Right: 2D spectral embedding with scaling
plt.subplot(1,2,2)
for i, node in enumerate(nodes):
    plt.scatter(X[i,0], X[i,1], color=color_map[labels[i]], s=200, edgecolor="k")
    plt.text(X[i,0]+0.02, X[i,1]+0.02, node, fontsize=12, weight="bold")

plt.title("2D Spectral Embedding (scaled by 1/sqrt(lambda))")
plt.xlabel("v2 / sqrt(lambda2)")
plt.ylabel("v3 / sqrt(lambda3)")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt

# Define edges
edges = [("a", "b"), ("b", "c"), ("d", "a"), ("a", "c"), ("c","d"),
         ("e", "f"), ("f", "g"), ("e", "g"), ("c", "e"), ("b", "e")]

# Create graph
G = nx.Graph()
G.add_edges_from(edges)

# Laplacian matrix
L = nx.laplacian_matrix(G).todense()
nodes = list(G.nodes())

# Eigen decomposition
eigvals, eigvecs = np.linalg.eigh(L)

print("Eigenvalues:")
print(np.round(eigvals, 3))

print("\nEigenvectors (each column corresponds to an eigenvector):")
for i in range(len(nodes)):
    print(f"Node {nodes[i]}: {np.round(eigvecs[i], 3)}")

# Spectral embedding in 2D using eigenvectors 2 and 3
embedding = eigvecs[:, 1:3]

# Plot embedding in 2D plane
plt.figure(figsize=(7, 6))
for i, node in enumerate(nodes):
    plt.scatter(embedding[i, 0], embedding[i, 1], s=200)
    plt.text(embedding[i, 0] + 0.02, embedding[i, 1] + 0.02, node, fontsize=12)

plt.title("2D Spectral Embedding (Eigenvectors 2 & 3)")
plt.xlabel("Eigenvector 2")
plt.ylabel("Eigenvector 3")
plt.grid(True)
plt.show()
```
