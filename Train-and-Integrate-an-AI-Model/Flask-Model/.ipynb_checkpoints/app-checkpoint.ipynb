```python
from flask import Flask, request, jsonify
import joblib

app = Flask(__name__)
model = joblib.load('movie_model.joblib')

genre_map = {'Action': 0, 'Comedy': 1, 'Drama': 2}

@app.route('/predict', methods=['POST'])
def predict():
    data = request.get_json()
    genre = data.get('genre')
    genre_code = genre_map.get(genre)
    if genre_code is None:
        return jsonify({'error': 'Invalid genre'}), 400
    prediction = model.predict([[genre_code]])
    return jsonify({'like': int(prediction[0])})

if __name__ == '__main__':
    app.run(port=5000)
```
