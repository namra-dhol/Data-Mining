```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
import joblib

# Training dataset
data = {
    'genre': ['Action', 'Comedy', 'Drama', 'Action', 'Comedy'],
    'like': [1, 0, 1, 0, 1]
}
df = pd.DataFrame(data)
df['genre'] = df['genre'].astype('category').cat.codes

# Features and target
X = df[['genre']]
y = df['like']

# Train model
model = DecisionTreeClassifier()
model.fit(X, y)

# Save model
joblib.dump(model, 'movie_model.joblib')
```
