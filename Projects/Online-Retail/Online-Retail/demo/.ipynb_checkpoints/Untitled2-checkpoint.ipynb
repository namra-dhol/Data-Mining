```python
import pandas as pd
import numpy as np
from mlxtend.frequent_patterns import apriori,association_rules
from itertools import combinations
```

```python
df = pd.read_csv("OnlineRetail.csv",encoding='ISO-8859-1')
df
```

```python
df.isnull().values.any()
```

```python
df.isnull().sum()
```

```python
df = df.dropna(axis=0,subset=['InvoiceNo'])
```

```python
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]
```

```python
df['Description'] = df['Description'].str.strip().str.lower()
```

```python
df.Country.unique()
```

```python
basket_france = (df[df['Country']=="France"].groupby(['InvoiceNo', 'Description'])['Quantity'].sum().unstack().reset_index().fillna(0)
                 .set_index('InvoiceNo'))
```

```python
basket_france
```

```python
def hot_encode(x):
    if(x<=0):
        return 0
    if(x>=1):
        return 1
```

```python
basket_encoded = basket_france.apply(lambda col: col.map(hot_encode))
```

```python
basket_encoded
```

```python
frq_items = apriori(basket_encoded, min_support=0.1, use_colnames=True)
rules = association_rules(frq_items,metric='lift',min_threshold=1)
rules = rules.sort_values(['confidence','lift'],ascending=[False,False])
```

```python
print(rules.head())
```

```python
transactions = []
for _, row in basket_encoded.iterrows():
    purchased_items = []
    
    for col in basket_encoded.columns:
        if row[col] == 1: 
            purchased_items.append(col)
    
    transactions.append(purchased_items)

def support(itemset):
    count = 0 
    for trans in transactions:
        if set(itemset).issubset(trans):
            count += 1
    return count / len(transactions)


def apriori_simple(transactions, min_support):
    items = sorted(set(i for trans in transactions for i in trans))
    freq_itemsets = []
    k = 1 # size of itemset
    current_items = items
    
    while current_items:
        candidate_sets = list(combinations(current_items, k))

        valid_sets = []
        for itemset in candidate_sets:
            sup = support(itemset)
            if sup >= min_support:
                freq_itemsets.append((set(itemset), sup))
                valid_sets.append(set(itemset))
        
        current_items = sorted(set.union(*valid_sets)) if valid_sets else []
        k += 1
    return freq_itemsets
```

```python
min_support = 0.1
frequent_itemsets = apriori_simple(transactions, min_support)
frequent_itemsets[len(frequent_itemsets)-1]
# print("Frequent Itemsets:")
# for i in frequent_itemsets:
#     print(i)
```
