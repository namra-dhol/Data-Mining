### Step 1: Load the Dataset


```python
import pandas as pd
```

```python
df = pd.read_csv('../OnlineRetail.csv', encoding='ISO-8859-1')
df
```

### Step 2: Drop the unnecessary records


```python
# Remove rows with missing CustomerID or Description
df = df.dropna(subset=['CustomerID', 'Description'])

# Remove cancelled transactions (InvoiceNo starting with 'C')
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]

# Remove negative or zero quantities
df = df[df['Quantity'] > 0]
```

### Step 4: Covert data to binary formate


```python
# Grouping by InvoiceNo and Description, summing quantities
basket = df.groupby(['InvoiceNo', 'Description'])['Quantity'].sum().unstack().fillna(0)

# Convert to binary
basket = basket.astype(bool).astype(int)
```

### Step 5: Define Apriori Function

```python
from itertools import combinations

def find_freqquent_itemsets(df, min_support):
    n = len(df)
    result = []

    for k in [1, 2, 3]:
        for items in combinations(df.columns, k):
            # mask = column wise list()
            mask = df[list(items)].all(axis=1)
            
            # support = (support_count / all transactions)
            support = mask.sum() / n
            
            if support >= min_support:
                result.append((frozenset(items), round(support, 2)))
                print(result)
        print("----------------------------------------------------")
    return result
    
```

### Step 6  Run Apriori

```python
df_items = df.drop(columns='InvoiceNo')
freqquent_itemsets = find_freqquent_itemsets(df_items, 1)

for itemset, support in freqquent_itemsets:
    print(f"{set(itemset)} -> support: {support}")
    
```
